```python
movies=movies[["genres","id","keywords","title","overview","production_companies","production_countries","release_date","spoken_languages"]]
movies.head(3)
```

```python
movies["genres"][0]
```

```python
import ast
def fetchname(obj):
    l=[]
    for di in ast.literal_eval(obj):
        l.append(di["name"])
    return l
```

```python
movies["genres"]=movies["genres"].apply(fetchname)
```

```python
movies["keywords"]=movies["keywords"].apply(fetchname)
```

```python
movies["overview"]=movies["overview"].apply(lambda x: x.split(" "))
```

```python
movies["production_companies"]= movies["production_companies"].apply(fetchname)
movies["production_countries"]= movies["production_countries"].apply(fetchname)
```

```python
movies["release_date"] = pd.to_datetime(movies["release_date"])
```

```python
movies["year"] = movies["release_date"].dt.year
movies["month"] = movies["release_date"].dt.month
movies["day"] = movies["release_date"].dt.day
movies=movies.drop(columns="release_date")
```

```python
def fetchcom(obj):
    count=0
    l=[]
    for dic in ast.literal_eval(obj):
        l.append(dic["name"])
        count=count+1
        if count==1:
            break 
    return l
   
```

```python
movies["production_companies"]=movies["production_companies"].apply(fetchcom)
```

```python
movies["production_countries"]=movies["production_countries"].apply(fetchcom)
```

```python
movies["spoken_languages"]=movies["spoken_languages"].apply(fetchcom)
```

```python
movies.dropna(inplace=True)
```

```python
movies["overview"]=movies["overview"].apply(lambda x:x.split(" "))
```

```python
movies.head(2)
```

```python
movies["date"]=movies[["year","month","day"]].values.tolist()
```

```python
movies
```

```python
movies["tag"]=movies["genres"]+movies["keywords"]+movies["overview"]+movies["production_companies"]+movies["production_countries"]+movies["spoken_languages"]+movies["date"]
```

```python
movies["tag"][0]
```

```python
vectors=movies[["id","title","tag"]]
```

```python
credits["crew"][0]
```

```python
def fetchar(obj):
    count=0
    l=[]
    for i in ast.literal_eval(obj):
            l.append(i["character"]) 
            l.append(i["name"])
            count=count+1
            if count==5:
                break
    return l
```

```python
def fetcrew(obj):
    l=[]
    for i in ast.literal_eval(obj):
        if i["job"]=="Director":
            l.append(i["name"])
    return l
```

```python
credits["cast"]=credits["cast"].apply(fetchar)
```

```python
credits["crew"]=credits["crew"].apply(fetcrew)
```

```python
credits.head(5)
```

```python
vectors["tag"]=vectors["tag"]+credits["cast"]+credits["crew"]
```

```python
vectors['tag'] = vectors['tag'].apply(
    lambda x: " ".join(map(str, x))
)
```

```python
vectors
```

```python
from sklearn.feature_extraction.text import CountVectorizer
vectorizer=CountVectorizer(stop_words="english",max_features=5000)
vector=vectorizer.fit_transform((vectors["tag"])).toarray()
```

```python
vectorizer.get_feature_names_out()
```

```python
vectors
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity=cosine_similarity(vector)
```

```python
similarity=pd.DataFrame(similarity)
```

```python
similarity
```

```python
movie_list=vectors["title"]
```

```python
def recommend(movie):
    
    index=vectors[vectors["title"]==movie].index[0]
    distance=similarity[index]
    distance=sorted(list(enumerate(distance)),reverse=True,key=lambda x:x[1])
    top=[]
    recommended_movies=[]
    for i in range(6):
        top.append(distance[i][0])
    for i in top:
        recommended_movies.append(vectors["title"].iloc[i])
    return recommended_movies
    
recommend("Avatar")
```
